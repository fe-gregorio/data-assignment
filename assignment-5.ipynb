```python
otherkin_df["gender_num"] = otherkin_df["Gender"].map({
    "Female": 0,
    "Male": 1
})

otherkin_df["therian_num"] = otherkin_df["Community"].map({
    "Therian": 1,
    "Otherkin": 0,
    "Vampire": 0
})
```

```python
values_df = otherkin_df[[
    "gender_num",
    "otherkin_num",
    "therian_num",
    "vampire_num"
]]

correlation_df = values_df.corr()

correlation_df
```

```python
quotes = otherkin_df.loc[:, "Identity origins":"Stories and cultural resources"].fillna("")

text = " ".join(quotes.values.flatten()).lower()

for symbol in ',.!?;:…[]()“”"|/—–':
    text = text.replace(symbol, " ")

words = text.split()
counts = pd.Series(words).value_counts()

counts.head()
```

```python
print(counts[counts > 1].to_string())
```

```python
import matplotlib.pyplot as plt

top_words = counts[counts > 1].head(30)

x = range(len(top_words))
#y = [1] * len(top_words)

plt.figure(figsize=(14, 6))
plt.scatter(x, y, s=top_words.values * 80, alpha=0.6)

for i, word in enumerate(top_words.index):
    plt.text(x[i], y[i], word, ha="center", va="center")

plt.axis("off")
plt.show()
```

```python
top_words = counts[counts > 1]

cols = 8
rows = (len(top_words) + cols - 1) // cols

plt.figure(figsize=(14, rows * 1.4))

for i, (word, count) in enumerate(top_words.items()):
    x = i % cols
    y = i // cols

    size = count * 180

    plt.scatter(x, y, s=size, alpha=0.6)
    plt.text(x, y, word, ha="center", va="center", fontsize=8)

plt.xlim(-0.7, cols - 0.3)
plt.ylim(rows - 0.3, -0.7)
plt.axis("off")
plt.show()
```

```python

import math
top_words = counts[counts > 1].head(100)

plt.figure(figsize=(10, 10))

for i, (word, count) in enumerate(top_words.items()):
    angle = i * 0.6
    radius = 0.35 * math.sqrt(i + 1)

    x = radius * math.cos(angle)
    y = radius * math.sin(angle)

    size = count * 250

    plt.scatter(x, y, s=size, alpha=0.6)
    plt.text(x, y, word, ha="center", va="center", fontsize=8)

plt.axis("off")
plt.axis("equal")
plt.show()

```
